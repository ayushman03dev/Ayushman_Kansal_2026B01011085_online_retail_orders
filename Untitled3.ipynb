```python
import pandas as pd
import numpy as np
from google.colab import files

# SETTINGS - change only if your file names are different
INPUT_FILE = "online_retail_orders_10000.csv"
OUTPUT_FILE = "online_retail_orders_cleaned.csv"

# We must initialize rows_removed before tracking removals
rows_removed = {}

print("STEP 1: Loading data...")
# Fixed: Use the INPUT_FILE variable instead of the Windows C:\ path
df = pd.read_csv(INPUT_FILE)
original_rows = len(df)
print("Rows:", df.shape[0], "| Columns:", df.shape[1])
print(df.head())

print("\nSTEP 2: Understanding the data...")
print(df.info())
print("\nSummary of number columns:")
print(df.describe())


print("\nSTEP 3: Cleaning column names...")
df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")
print(list(df.columns))


print("\nSTEP 4: Cleaning text columns...")
text_columns = ["customer_segment", "region", "state", "city",
                "sales_channel", "category", "product"]

for col in text_columns:
    df[col] = df[col].astype(str).str.strip()        # remove spaces
    df[col] = df[col].str.replace(r"\s+", " ", regex=True)  # many spaces -> one
    df[col] = df[col].str.title()                    # Title Case

# Put back real "missing" values (astype(str) turns NaN into the word "Nan")
df[text_columns] = df[text_columns].replace("Nan", np.nan)

# IDs should be UPPERCASE and without spaces
df["order_id"] = df["order_id"].astype(str).str.strip().str.upper()
df["customer_id"] = df["customer_id"].astype(str).str.strip().str.upper()
print("Text columns cleaned.")

print("\nSTEP 5: Fixing data types...")
df["order_date"] = pd.to_datetime(df["order_date"], errors="coerce")

number_columns = ["quantity", "unit_price_usd", "discount_pct",
                  "shipping_days", "revenue_usd"]
for col in number_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")

print(df.dtypes)

print("\nSTEP 6: Checking missing values...")
print(df.isnull().sum())

# 6a. Rows with no order_id, customer_id or order_date are useless -> drop
before = len(df)
df = df.dropna(subset=["order_id", "customer_id", "order_date"])
rows_removed["Missing id/date"] = before - len(df)

# 6b. Missing text -> "Unknown"
for col in text_columns:
    df[col] = df[col].fillna("Unknown")

# 6c. Missing numbers -> fill with the median (middle value)
for col in ["quantity", "unit_price_usd", "discount_pct", "shipping_days"]:
    median_value = df[col].median()
    df[col] = df[col].fillna(median_value)

print("Missing values left:", df.isnull().sum().sum() - df["revenue_usd"].isnull().sum(),
      "(revenue is recalculated in Step 9)")

print("\nSTEP 7: Removing duplicates...")
before = len(df)
df = df.drop_duplicates()                          # fully identical rows
df = df.drop_duplicates(subset="order_id")         # same order id twice
rows_removed["Duplicates"] = before - len(df)
print("Duplicates removed:", rows_removed["Duplicates"])

print("\nSTEP 8: Removing impossible values...")
before = len(df)

valid_rows = (
    (df["quantity"] > 0) &
    (df["unit_price_usd" ] > 0) &
    (df["discount_pct"] >= 0) & (df["discount_pct"] <= 100) &
    (df["shipping_days"] >= 0)
)
df = df[valid_rows]
rows_removed["Impossible values"] = before - len(df)
print("Rows with impossible values removed:", rows_removed["Impossible values"])

print("\nSTEP 9: Checking revenue...")
df["revenue_check"] = (
    df["quantity"] * df["unit_price_usd"] * (1 - df["discount_pct"] / 100)
).round(2)

# Count rows where the given revenue differs from our calculation by more than 2 cents
wrong = np.abs(df["revenue_check"] - df["revenue_usd"]) > 0.02
print("Rows with wrong revenue:", wrong.sum())

# Use the calculated value wherever the old one was wrong or missing
df["revenue_usd"] = np.where(wrong | df["revenue_usd"].isnull(),
                             df["revenue_check"], df["revenue_usd"])
df = df.drop(columns="revenue_check")

print("\nSTEP 10: Looking for outliers in revenue...")
q1 = np.percentile(df["revenue_usd"], 25)
q3 = np.percentile(df["revenue_usd"], 75)
iqr = q3 - q1
lower_limit = q1 - 1.5 * iqr
upper_limit = q3 + 1.5 * iqr

df["is_revenue_outlier"] = (df["revenue_usd"] < lower_limit) | (df["revenue_usd"] > upper_limit)
print("Normal range: %.2f to %.2f" % (lower_limit, upper_limit))
print("Outlier orders flagged:", df["is_revenue_outlier"].sum())


print("\nSTEP 11: Adding new columns...")
df["order_year"] = df["order_date"].dt.year
df["order_month"] = df["order_date"].dt.month
df["month_name"] = df["order_date"].dt.month_name()
df["day_of_week"] = df["order_date"].dt.day_name()
df["gross_amount_usd"] = (df["quantity"] * df["unit_price_usd"]).round(2)
df["discount_amount_usd"] = (df["gross_amount_usd"] - df["revenue_usd"]).round(2)

# Make whole-number columns proper integers
df["quantity"] = df["quantity"].astype(int)
df["discount_pct"] = df["discount_pct"].astype(int)
df["shipping_days"] = df["shipping_days"].astype(int)


df = df.sort_values("order_date").reset_index(drop=True)

print("\nSTEP 12: Final summary")
print("Rows before cleaning:", original_rows)
print("Rows after cleaning :", len(df))
for reason, count in rows_removed.items():
    print("  removed (%s): %d" % (reason, count))
print("Missing values left :", df.isnull().sum().sum())
print("Duplicate rows left :", df.duplicated().sum())

df.to_csv(OUTPUT_FILE, index=False)
print("\nCleaned data saved to:", OUTPUT_FILE)

# Fixed: Download using the defined OUTPUT_FILE variable
files.download(OUTPUT_FILE)
```
